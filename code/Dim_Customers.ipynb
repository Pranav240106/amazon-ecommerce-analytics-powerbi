```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_excel("D:\Projects\Amazon Data Analysis\output\Amazon Cleaned Data.xlsx")
```

```python
dim_customer = df[['CustomerID', 'CustomerName']].copy()
```

```python
dim_customer = dim_customer.drop_duplicates(subset='CustomerID')
```

```python
print("Total Rows :", len(dim_customer))
print("Unique CustomerID :", dim_customer['CustomerID'].nunique())
```

```python
dim_customer.to_excel("Dim_Customer.xlsx", index=False)
```

```python

dim_product = (
    df[['ProductID', 'ProductName', 'Category', 'Brand']]
    .drop_duplicates(subset='ProductID')
    .reset_index(drop=True)
)

print("Rows:", len(dim_product))
print("Unique ProductID:", dim_product['ProductID'].nunique())

dim_product.to_excel("Dim_Product.xlsx", index=False)
```

```python
# Convert OrderDate to datetime
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

# Create Date Dimension
dim_date = pd.DataFrame({
    'Date': sorted(df['OrderDate'].drop_duplicates())
})

dim_date['Year'] = dim_date['Date'].dt.year
dim_date['Quarter'] = 'Q' + dim_date['Date'].dt.quarter.astype(str)
dim_date['Month Number'] = dim_date['Date'].dt.month
dim_date['Month Name'] = dim_date['Date'].dt.strftime('%B')
dim_date['Day'] = dim_date['Date'].dt.day
dim_date['Weekday'] = dim_date['Date'].dt.day_name()

print("Rows:", len(dim_date))
print("Unique Dates:", dim_date['Date'].nunique())

dim_date.to_excel("Dim_Date.xlsx", index=False)
```
